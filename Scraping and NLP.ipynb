```python
# import libraries
import requests
from bs4 import BeautifulSoup
import re
from collections import Counter
```

```python
# URL of the Project Gutenberg novel
url = "https://www.gutenberg.org/files/2701/2701-h/2701-h.htm"

# Send a request to the website
response = requests.get(url)

# Store the HTML content
html_content = response.text

# Just to check if it worked
print("Downloaded characters:", len(html_content))
```

```python
# STEP 2: Parse HTML and extract text
soup = BeautifulSoup(html_content, "html.parser")

# Get all text from the page
text = soup.get_text()

# Check if text was extracted
print(text) 
```

```python
with open("novel.txt", "w", encoding="utf-8") as f:
    f.write(text)

print("Whole website text saved to novel.txt")
```

```python
# STEP 3: Clean the text
text = text.lower()                     # lowercase
text = re.sub(r'[^a-z\s]', '', text)    # remove punctuation & numbers
words = text.split()                    # split into words

# Check result
print(words)
print("Total words:", len(words))
```

```python
with open("all_words.txt", "w", encoding="utf-8") as f:
    for word in words:
        f.write(word + "\n")

print("All words saved to all_words.txt")
```

```python
print(len(words))
```

```python

# STEP 4: Count word frequency
word_counts = Counter(words)

# Show the 20 most common words
most_common_words = word_counts.most_common(20)
print(most_common_words)
```
